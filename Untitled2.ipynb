```python
import numpy as np
import pandas as pd
import torch
from transformers import pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
import matplotlib.pyplot as plt
```

```python
# Load the sentiment analysis pipeline from Hugging Face
sentiment_pipeline = pipeline('sentiment-analysis')

# Load the text classification pipeline (e.g., 20 newsgroups topics)
classification_pipeline = pipeline('zero-shot-classification')
```

```python
def analyze_post(user_input):
    # Sentiment analysis
    sentiment = sentiment_pipeline(user_input)

    # Topic classification (using zero-shot classification)
    candidate_labels = ['politics', 'sports', 'technology', 'health', 'entertainment']
    topic = classification_pipeline(user_input, candidate_labels=candidate_labels)

    return sentiment, topic
```

```python
# Get user input and analyze the post
user_input = input("Enter a social media post: ")
sentiment, topic = analyze_post(user_input)

print(f"Sentiment: {sentiment[0]['label']} with score: {sentiment[0]['score']}")
print(f"Predicted Topic: {topic['labels'][0]} with confidence: {topic['scores'][0]}")
```

```python
# Example of visualizing sentiment distribution for multiple posts
posts = ["I love technology!", "Politics is so frustrating", "The new movie was amazing!", "I hate waiting in line."]
sentiments = [sentiment_pipeline(post)[0]['label'] for post in posts]

# Plot the sentiment distribution
sentiment_counts = pd.Series(sentiments).value_counts()
sentiment_counts.plot(kind='bar', color=['green', 'red', 'gray'])
plt.title("Sentiment Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Frequency")
plt.show()
```
